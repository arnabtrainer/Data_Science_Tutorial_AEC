## 5. Mathematical or formal lens

For independent observations with variance \(\sigma^2\), the mean has standard error \(\sigma/\sqrt{n}\). Under broad conditions its sampling distribution approaches Normal.

````python
rng_local = np.random.default_rng(42)
population = rng_local.exponential(scale=10, size=200_000)

def sampling_means(sample_size: int, repetitions: int = 5000) -> np.ndarray:
    return rng_local.choice(population, size=(repetitions, sample_size), replace=True).mean(axis=1)

means_5 = sampling_means(5)
means_50 = sampling_means(50)
print("Population mean:", population.mean())
print("SD of means n=5 :", means_5.std())
print("SD of means n=50:", means_50.std())

fig, ax = plt.subplots(figsize=(7, 4))
ax.hist(means_50, bins=35, edgecolor="black")
ax.set(title="Sampling distribution of the mean (n=50)", xlabel="Sample mean", ylabel="Frequency")
plt.show()
````
