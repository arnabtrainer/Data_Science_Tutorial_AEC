# Phase 4: Statistics and Mathematics

## Phase goal

Understand the probability, inference, linear algebra, calculus, and optimization that make data science rigorous.

**Estimated lesson time:** 17.6 hours  
**Recommended additional practice:** 12 hours  
**Prerequisite:** Basic algebra and Python arrays; every required statistical idea is developed here.

This phase progresses from conceptual foundations to practitioner execution and advanced failure analysis. Complete lessons in order unless a diagnostic assessment shows that you already satisfy the readiness criteria.

## Phase learning outcomes

After completing this phase, you should be able to:

1. Explain every central concept without relying on library vocabulary alone.
2. Implement the core mechanisms in small, testable examples.
3. Select methods according to assumptions, constraints, and decision cost.
4. Detect common correctness, leakage, interpretation, and operational failures.
5. Complete the phase lab without copying the reference solution.
6. Produce the phase project with evidence, tests, limitations, and a professional handoff.

## How to study this phase

Use a three-pass method:

### Pass 1 — Understand
Read the mental model, vocabulary, and formal lens. Run the example only after predicting its result.

### Pass 2 — Reconstruct
Close the example and recreate the mechanism in a fresh cell or file. Use assertions and edge cases.

### Pass 3 — Transfer
Apply the method to a bundled dataset or a new scenario. Explain assumptions, limitations, and decision impact.

A suggested cadence is **five hours per week for approximately 4 week(s)** for the core lessons, followed by the lab, assessment, and project.

## Phase lab

        **Scenario:** A team compares a redesigned process with the existing process using completion-time samples.

        **Deliverable:** A statistically defensible analysis including design, uncertainty, test, effect size, assumptions, and practical recommendation.

        Required work:

        - Define population, estimand, null, alternative, and minimum meaningful effect.
- Visualize distributions and inspect independence/design assumptions.
- Calculate descriptive summaries and confidence intervals.
- Run an appropriate comparison test.
- Calculate standardized and raw effect sizes.
- Perform a sensitivity or bootstrap analysis.
- Separate statistical and practical conclusions.

        Files:

        - `../exercises/Phase_04_Lab.ipynb`
        - `../solutions/Phase_04_Lab_Solution.ipynb`
