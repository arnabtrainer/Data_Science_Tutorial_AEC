## 5. Mathematical or formal lens

Gradient descent updates \(\theta_{t+1}=\theta_t-\eta\nabla_\theta L(\theta_t)\). The negative gradient is the steepest local descent direction under Euclidean geometry.

### Interpretation discipline

## 7. Worked Python example

Read the code once without running it. Predict:

- the important intermediate objects and their shapes/types;

Then run the cell and reconcile the actual output with your prediction.

````python
x = np.array([1., 2., 3., 4., 5.])
y = np.array([3., 5., 7., 9., 11.])
w, b = 0.0, 0.0
learning_rate = 0.02
history = []

for step in range(500):
    prediction = w*x + b
    error = prediction - y
    loss = np.mean(error**2)
    dw = 2*np.mean(error*x)
    db = 2*np.mean(error)
    w -= learning_rate*dw
    b -= learning_rate*db
    history.append(loss)

print(f"Learned w={w:.4f}, b={b:.4f}, final loss={history[-1]:.8f}")
fig, ax = plt.subplots(figsize=(6, 4))
ax.plot(history)
ax.set(title="Gradient descent reduces the objective", xlabel="Iteration", ylabel="MSE")
ax.set_yscale("log")
plt.show()
````
