## 4. First-principles workflow

        1. State the input, expected output, and invalid cases before writing code.
2. Choose the simplest data representation that preserves the domain meaning.
3. Separate calculation from input/output so logic can be tested independently.
4. Make state changes explicit; avoid hidden globals and accidental aliasing.
5. Exercise the normal path, boundaries, and failures before declaring the program complete.

        ### Before using a library

        For the central operation, write down:

        - the shape and meaning of every input;
        - the expected output shape and unit;
        - the objective or invariant;
        - one hand-calculated example;
        - computational complexity at a high level;
        - one condition under which the method becomes invalid.

        Library fluency is valuable only after the contract is understood.

## 5. Mathematical or formal lens

A useful test follows Arrange–Act–Assert and checks one observable contract. Test behaviour and edge cases, not implementation trivia.

### Interpretation discipline

A formula does not establish that its assumptions are true. A successful function call does not establish that its output is meaningful. Always separate:

- **calculation correctness** — did the code implement the formula or contract?
- **statistical validity** — does the design justify the inference?
- **operational validity** — will equivalent inputs exist at decision time?
- **decision validity** — is the output useful under real costs and constraints?

````python
from typing import Sequence

def normalize(values: Sequence[float]) -> list[float]:
    """Min-max normalize a nonconstant sequence."""
    if len(values) == 0:
        raise ValueError("values cannot be empty")
    minimum, maximum = min(values), max(values)
    if minimum == maximum:
        raise ValueError("values must not all be identical")
    return [(value - minimum) / (maximum - minimum) for value in values]

def test_normalize_boundaries():
    result = normalize([10, 15, 20])
    assert result[0] == 0.0
    assert result[-1] == 1.0
    assert len(result) == 3

test_normalize_boundaries()
print(normalize([10, 15, 20]))
print("Lightweight unit test passed.")
````

## 10. Common mistakes and safeguards

        | Common mistake | Professional safeguard |
        |---|---|
        | Relying on notebook state | Restart the kernel and run all cells in order before trusting the result. |
| Catching every exception | Catch only recoverable errors; preserve traceback and context for unexpected failures. |
| Using mutable defaults | Use `None` or `default_factory` so calls/instances do not share accidental state. |
| Writing large multipurpose functions | Separate validation, transformation, I/O, and orchestration. |
