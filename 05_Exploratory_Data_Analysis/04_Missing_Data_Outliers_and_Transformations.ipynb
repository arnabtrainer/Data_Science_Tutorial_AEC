A formula does not establish that its assumptions are true. A successful function call does not establish that its output is meaningful. Always separate:

- **calculation correctness** — did the code implement the formula or contract?
- **statistical validity** — does the design justify the inference?
- **operational validity** — will equivalent inputs exist at decision time?
- **decision validity** — is the output useful under real costs and constraints?

## 7. Worked Python example

````python
df = pd.read_csv(DATA_DIR / "house_prices.csv")
q1, q3 = df["price"].quantile([.25,.75])
iqr = q3-q1
lower, upper = q1-1.5*iqr, q3+1.5*iqr
outlier_mask = ~df["price"].between(lower, upper)

analysis = pd.Series({
    "missing_school_rating": int(df["school_rating"].isna().sum()),
    "price_outliers_iqr_rule": int(outlier_mask.sum()),
    "price_skew_before": float(df["price"].skew()),
    "price_skew_after_log1p": float(np.log1p(df["price"]).skew()),
})
display(analysis.to_frame("value"))

# Preserve an indicator before imputation when missingness may be informative.
df["school_rating_missing"] = df["school_rating"].isna().astype(int)
df["school_rating_filled"] = df["school_rating"].fillna(df["school_rating"].median())
display(df[["school_rating","school_rating_missing","school_rating_filled"]].head(12))
````
