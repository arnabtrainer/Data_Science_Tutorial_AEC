# Phase 5: Exploratory Data Analysis

## Phase goal

Interrogate data quality and structure before modelling, while avoiding leakage, bias, and unsupported causal claims.

**Estimated lesson time:** 11.6 hours  
**Recommended additional practice:** 8 hours  
**Prerequisite:** Pandas, visualization, and descriptive-statistics foundations.

This phase progresses from conceptual foundations to practitioner execution and advanced failure analysis. Complete lessons in order unless a diagnostic assessment shows that you already satisfy the readiness criteria.

## Phase learning outcomes

After completing this phase, you should be able to:

1. Explain every central concept without relying on library vocabulary alone.
2. Implement the core mechanisms in small, testable examples.
3. Select methods according to assumptions, constraints, and decision cost.
4. Detect common correctness, leakage, interpretation, and operational failures.
5. Complete the phase lab without copying the reference solution.
6. Produce the phase project with evidence, tests, limitations, and a professional handoff.

## How to study this phase

Use a three-pass method:

### Pass 1 — Understand
Read the mental model, vocabulary, and formal lens. Run the example only after predicting its result.

### Pass 2 — Reconstruct
Close the example and recreate the mechanism in a fresh cell or file. Use assertions and edge cases.

### Pass 3 — Transfer
Apply the method to a bundled dataset or a new scenario. Explain assumptions, limitations, and decision impact.

A suggested cadence is **five hours per week for approximately 3 week(s)** for the core lessons, followed by the lab, assessment, and project.

## Phase lab

        **Scenario:** A retention team needs an exploratory report before approving churn modelling.

        **Deliverable:** A reproducible EDA notebook with framing, quality audit, target analysis, segment findings, leakage review, and modelling recommendations.

        Required work:

        - Define the observation unit, target horizon, and intervention decision.
- Create a data dictionary and quality audit.
- Quantify target prevalence and uncertainty.
- Compare numeric and categorical features by target.
- Inspect missingness, outliers, interactions, and segments.
- Create a feature-availability/leakage table.
- Conclude with limitations and a modelling protocol.

        Files:

        - `../exercises/Phase_05_Lab.ipynb`
        - `../solutions/Phase_05_Lab_Solution.ipynb`
        - `../assessments/Phase_05_Assessment.ipynb`
