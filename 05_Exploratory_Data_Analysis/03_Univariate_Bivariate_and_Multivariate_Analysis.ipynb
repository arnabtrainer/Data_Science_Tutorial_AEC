## 5. Mathematical or formal lens

A marginal relationship can reverse after conditioning on a third variable; always inspect meaningful segments and sample sizes.

### Interpretation discipline

A formula does not establish that its assumptions are true. A successful function call does not establish that its output is meaningful. Always separate:

- **calculation correctness** — did the code implement the formula or contract?
- **statistical validity** — does the design justify the inference?
- **operational validity** — will equivalent inputs exist at decision time?
- **decision validity** — is the output useful under real costs and constraints?

````python
df = pd.read_csv(DATA_DIR / "customer_churn.csv")
numeric_summary = df.groupby("churn")[["tenure_months", "monthly_charges", "support_tickets_90d", "weekly_usage_hours"]].agg(["mean","median"])
display(numeric_summary.round(2))

rates = pd.crosstab(df["contract_type"], df["churn"], normalize="index").rename(columns={0:"retained_rate",1:"churn_rate"})
display(rates.round(3))

fig, ax = plt.subplots(figsize=(7, 4))
ax.hist([df.loc[df.churn==0,"monthly_charges"], df.loc[df.churn==1,"monthly_charges"]],
        bins=25, label=["Retained","Churned"], alpha=.65)
ax.set(title="Monthly charges by outcome", xlabel="Monthly charges", ylabel="Customers")
ax.legend()
plt.show()
````
