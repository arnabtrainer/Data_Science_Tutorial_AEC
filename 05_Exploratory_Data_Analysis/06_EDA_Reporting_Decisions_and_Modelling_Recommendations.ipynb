## 4. First-principles workflow

        1. Frame the decision, row grain, target, and time boundary before opening charts.
2. Audit provenance and quality before trusting descriptive statistics.
3. Move from one variable to relationships and then meaningful segments.
4. Challenge every strong pattern for leakage, confounding, selection, or data defects.
5. Close with evidence, limitations, modelling implications, and ownership.

        ### Before using a library

        - one condition under which the method becomes invalid.

        Library fluency is valuable only after the contract is understood.

## 5. Mathematical or formal lens

A strong finding states the population, direction, magnitude, uncertainty, and limitation—not only a chart or correlation.

### Interpretation discipline

A formula does not establish that its assumptions are true. A successful function call does not establish that its output is meaningful. Always separate:

- **calculation correctness** — did the code implement the formula or contract?
- **statistical validity** — does the design justify the inference?
- **operational validity** — will equivalent inputs exist at decision time?
- **decision validity** — is the output useful under real costs and constraints?

## 10. Common mistakes and safeguards

        | Common mistake | Professional safeguard |
        |---|---|
        | EDA without a decision boundary | Define unit, target, cutoff, and outcome horizon first. |
| Treating data-cleaning choices as neutral | Document how each treatment changes the analysis population. |
| Over-interpreting correlations | Inspect confounders, segments, timing, and alternative explanations. |
| Leaking future information | Create an availability table for every candidate feature. |
| Reporting charts without implications | Connect each finding to risk, modelling, or action. |
