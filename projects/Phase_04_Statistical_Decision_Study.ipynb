# Phase 4 Project: Statistical Decision Study

**Phase:** Statistics and Mathematics  
**Difficulty:** Practitioner-to-Advanced  
**Purpose:** Design and analyze a controlled comparison with uncertainty intervals, tests, effect size, assumptions, and practical interpretation.

This project is deliberately open-ended. The goal is not to reproduce a hidden answer but to make defensible technical and analytical decisions.

## Scenario

        A team compares a redesigned process with the existing process using completion-time samples.

        ## Required deliverable

        A statistically defensible analysis including design, uncertainty, test, effect size, assumptions, and practical recommendation.

        ## Functional requirements

        - [ ] Define population, estimand, null, alternative, and minimum meaningful effect.
- [ ] Visualize distributions and inspect independence/design assumptions.
- [ ] Calculate descriptive summaries and confidence intervals.
- [ ] Run an appropriate comparison test.
- [ ] Calculate standardized and raw effect sizes.
- [ ] Perform a sensitivity or bootstrap analysis.
- [ ] Separate statistical and practical conclusions.

## Required report sections

1. Executive summary
2. Decision and scope
3. Data contract, lineage, and quality
4. Method and first-principles rationale
5. Baseline and alternatives
6. Evaluation design
7. Results with uncertainty and slices
8. Error/failure analysis
9. Limitations and excluded uses
10. Recommendation and next experiment
11. Reproduction instructions
12. Artifact manifest
