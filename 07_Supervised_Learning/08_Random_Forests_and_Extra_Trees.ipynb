## 5. Mathematical or formal lens

### Interpretation discipline

- **calculation correctness** — did the code implement the formula or contract?
- **statistical validity** — does the design justify the inference?

````python
from sklearn.datasets import load_breast_cancer
from sklearn.ensemble import ExtraTreesClassifier, RandomForestClassifier
from sklearn.model_selection import cross_val_score

X,y=load_breast_cancer(return_X_y=True)
models={
    "random_forest":RandomForestClassifier(n_estimators=200,min_samples_leaf=3,random_state=42,n_jobs=1),
    "extra_trees":ExtraTreesClassifier(n_estimators=200,min_samples_leaf=3,random_state=42,n_jobs=1),
}
for name,model in models.items():
    scores=cross_val_score(model,X,y,cv=5,scoring="roc_auc")
    print(name,round(scores.mean(),4),round(scores.std(),4))
````
