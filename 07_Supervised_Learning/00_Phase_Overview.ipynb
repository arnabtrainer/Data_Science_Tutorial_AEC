# Phase 7: Supervised Learning

## Phase goal

Master regression and classification algorithms from first principles through tuned, explainable ensembles.

**Estimated lesson time:** 25.5 hours  
**Recommended additional practice:** 17 hours  
**Prerequisite:** Machine-learning experiment design, preprocessing pipelines, and metrics.

This phase progresses from conceptual foundations to practitioner execution and advanced failure analysis. Complete lessons in order unless a diagnostic assessment shows that you already satisfy the readiness criteria.

## Phase learning outcomes

After completing this phase, you should be able to:

1. Explain every central concept without relying on library vocabulary alone.
2. Implement the core mechanisms in small, testable examples.
3. Select methods according to assumptions, constraints, and decision cost.
4. Detect common correctness, leakage, interpretation, and operational failures.
5. Complete the phase lab without copying the reference solution.
6. Produce the phase project with evidence, tests, limitations, and a professional handoff.

## How to study this phase

Use a three-pass method:

### Pass 1 — Understand
Read the mental model, vocabulary, and formal lens. Run the example only after predicting its result.

### Pass 2 — Reconstruct
Close the example and recreate the mechanism in a fresh cell or file. Use assertions and edge cases.

### Pass 3 — Transfer
Apply the method to a bundled dataset or a new scenario. Explain assumptions, limitations, and decision impact.

A suggested cadence is **five hours per week for approximately 6 week(s)** for the core lessons, followed by the lab, assessment, and project.

## Phase lab

        **Scenario:** A modelling review compares algorithms for house-price regression and customer-churn classification.

        **Deliverable:** Two tuned, explainable modelling pipelines with identical validation protocols and business-facing recommendations.

- Use nested or carefully separated tuning and evaluation.
- Report uncertainty, residual/error slices, and calibration.
- Use held-out permutation importance or equivalent explanations.
- Perform sensitivity analysis on important hyperparameters.
- Create model cards with intended and excluded uses.
