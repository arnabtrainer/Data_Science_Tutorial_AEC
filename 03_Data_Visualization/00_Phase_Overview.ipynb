# Phase 3: Data Visualization

## Phase goal

Choose truthful visual encodings and turn analysis into clear, decision-oriented stories.

**Estimated lesson time:** 10.0 hours  
**Recommended additional practice:** 7 hours  
**Prerequisite:** Be comfortable loading and transforming DataFrames.

This phase progresses from conceptual foundations to practitioner execution and advanced failure analysis. Complete lessons in order unless a diagnostic assessment shows that you already satisfy the readiness criteria.

## Phase learning outcomes

After completing this phase, you should be able to:

1. Explain every central concept without relying on library vocabulary alone.
2. Implement the core mechanisms in small, testable examples.
3. Select methods according to assumptions, constraints, and decision cost.
4. Detect common correctness, leakage, interpretation, and operational failures.
5. Complete the phase lab without copying the reference solution.
6. Produce the phase project with evidence, tests, limitations, and a professional handoff.

## How to study this phase

Use a three-pass method:

### Pass 1 — Understand
Read the mental model, vocabulary, and formal lens. Run the example only after predicting its result.

### Pass 2 — Reconstruct
Close the example and recreate the mechanism in a fresh cell or file. Use assertions and edge cases.

### Pass 3 — Transfer
Apply the method to a bundled dataset or a new scenario. Explain assumptions, limitations, and decision impact.

A suggested cadence is **five hours per week for approximately 2 week(s)** for the core lessons, followed by the lab, assessment, and project.

## Phase lab

        **Scenario:** An executive needs to understand where sales growth and profit are coming from and where margin risk is emerging.

        **Deliverable:** A six-chart visual narrative with a one-page executive summary and accessibility notes.

        Required work:

        - Write the decision question for each chart before plotting.
- Create trend, category comparison, distribution, and relationship views.
- Annotate the most decision-relevant observation.
- Show data coverage and sample size.
- Avoid dual axes and unjustified truncated baselines.
- Write one evidence-limitation-action statement per visual.
- Review labels, legibility, and accessibility.

        - `../exercises/Phase_03_Lab.ipynb`
        - `../solutions/Phase_03_Lab_Solution.ipynb`
