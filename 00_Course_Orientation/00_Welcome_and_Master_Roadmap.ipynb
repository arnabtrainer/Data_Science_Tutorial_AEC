# Data Science, Machine Learning, Deep Learning, and MLOps Master Course

Welcome to an offline-first, from-scratch-to-advanced learning program.

This package is designed as a **complete self-study system**, not a collection of disconnected examples. It integrates:

- conceptual explanations and first-principles reasoning;
- mathematics with numerical intuition;
- executable Python programs;
- guided and independent exercises;
- separate reference solutions;
- assessments and answer keys;
- phase projects and worked capstones;
- production engineering, deployment, monitoring, and governance.

## Ten phases

1. Python Foundation
2. NumPy and Pandas
3. Data Visualization
4. Statistics and Mathematics
5. Exploratory Data Analysis
6. Machine Learning Fundamentals
7. Supervised Learning
8. Unsupervised Learning
9. Deep Learning
10. Real-World ML and MLOps

## What “from scratch to advanced” means here

**Foundation** notebooks define the vocabulary and build small mechanisms carefully.

**Practitioner** notebooks apply those mechanisms to realistic datasets, compare alternatives, validate assumptions, and interpret results.

**Advanced** notebooks derive objectives, expose failure modes, address reproducibility and scalability, and connect experiments to production systems.

The progression is:

```text
Understand the decision
        ↓
Understand and validate data
        ↓
Calculate small examples manually
        ↓
Implement the mechanism
        ↓
Use professional libraries
        ↓
Evaluate generalization and uncertainty
        ↓
Analyze failure modes
        ↓
Package, deploy, monitor, and govern
```

## Recommended learning tracks

### Complete track
Follow every lesson, lab, assessment, project, and capstone. Best for a learner beginning from first principles.

### Data Analyst → Data Scientist track
Complete Phases 1–6, then regression/classification in Phase 7, selected Phase 8 lessons, and production fundamentals in Phase 10.

### Machine Learning Engineer track
Complete Phases 1–7, Phase 9, and all of Phase 10. Do not skip EDA, statistics, or evaluation design.

### Deep Learning track
Complete Python, NumPy, mathematics, ML fundamentals, all of Phase 9, and deployment/monitoring in Phase 10.

### Revision/interview track
Use the phase overviews, assessments, formula sheet, glossary, interview bank, and worked capstones. Reconstruct examples rather than merely rereading them.

## Master completion standard

A phase is complete when you can:

- [ ] explain the core ideas in ordinary language and notation;
- [ ] reproduce the central calculations without copying;
- [ ] implement the method with clean, tested Python;
- [ ] state assumptions and failure modes;
- [ ] choose defensible evaluation metrics and splits;
- [ ] interpret outputs without overclaiming;
- [ ] complete the lab and score at least 80% on the assessment;
- [ ] produce the phase project at portfolio standard.

A model running successfully is not evidence that it is correct, useful, fair, or deployable. This distinction is a recurring theme throughout the course.
