# Notation, Coding, and Reporting Conventions

## Mathematical notation

| Symbol | Meaning |
|---|---|
| \(x_i\) | Feature vector or observation i |
| \(y_i\) | Observed target for observation i |
| \(\hat{y}_i\) | Prediction |
| \(\theta\) | Generic model parameters |
| \(w, b\) | Weights and bias/intercept |
| \(X\) | Design matrix: rows are observations, columns are features |
| \(L(\theta)\) | Loss/objective as a function of parameters |
| \(E[\cdot]\) | Expected value |
| \(P(A\mid B)\) | Conditional probability of A given B |
| \(\nabla_\theta L\) | Gradient of loss with respect to parameters |

Shapes and units should be stated whenever they are not obvious.

## Coding conventions

- Prefer descriptive names over abbreviations.
- Separate I/O, validation, transformation, modelling, and orchestration.
- Use `pathlib.Path` for files.
- Preserve raw data and write outputs under `artifacts/`.
- Use pipelines for learned preprocessing.
- Set random seeds and record versions, but understand that determinism can have platform limits.
- Add assertions for invariants, not as a substitute for boundary validation in production.
- Keep notebook cells restartable and avoid hidden state.

## Reporting conventions

Every quantitative claim should identify:

1. population and data coverage;
2. metric/estimand and units;
3. comparison or baseline;
4. uncertainty or sample size;
5. assumptions and important exclusions;
6. whether the claim is descriptive, predictive, or causal;
7. decision implication.

Example:

> On the held-out 2025 customer cohort, the pipeline achieved a PR-AUC of X against a prevalence baseline of Y. Performance was lower for segment Z. The result supports ranking for review but not automatic denial or intervention effectiveness.
