# How to Learn with These Notebooks

Passive reading produces familiarity, not reliable skill. Use each notebook as an experiment.

## The prediction–execution–explanation cycle

1. **Predict:** Before running code, write what you expect and why.
2. **Execute:** Run the smallest relevant cell.
3. **Observe:** Record output, shape, units, and anomalies.
4. **Explain:** Connect the output to mechanism and assumptions.
5. **Vary:** Change one input or parameter at a time.
6. **Test:** Add an assertion or edge case.
7. **Transfer:** Apply the idea to another dataset.

## Three levels of mastery

| Level | Evidence |
|---|---|
| Recognition | You understand an explanation while reading it |
| Recall | You can explain and reconstruct it without notes |
| Transfer | You can select, adapt, validate, and defend it in a new problem |

The course targets **transfer**. A notebook is not complete because all cells are green.

## How to use solutions

1. Spend a defined, serious attempt on the exercise.
2. Reduce the problem to a smaller example.
3. Revisit the relevant concept—not the finished solution.
4. Open the solution only after documenting where you are blocked.
5. Compare contracts, assumptions, tests, and decomposition.
6. Close the solution and rebuild independently.

Copying a reference solution creates an illusion of progress. Reconstruction creates skill.

## Weekly review

At the end of each week:

- [ ] Reproduce one algorithm or transformation from a blank notebook.
- [ ] Explain three topics aloud at beginner and interview depth.
- [ ] Revisit every failed assessment item.
- [ ] Refactor one earlier program for clarity and tests.
- [ ] Add one item to your personal failure-mode checklist.
- [ ] Update `reference/PROGRESS_TRACKER.csv`.

Use spaced review after approximately 1 day, 1 week, 1 month, and 3 months.
