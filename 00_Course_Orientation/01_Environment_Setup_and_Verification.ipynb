# Environment Setup and Verification

## Recommended environment

- Python 3.11
- JupyterLab 4
- A virtual environment or Conda environment dedicated to this course

### `venv` installation

```bash
python -m venv .venv

# Windows PowerShell
.venv\Scripts\Activate.ps1

# macOS/Linux
source .venv/bin/activate

python -m pip install --upgrade pip
python -m pip install -r requirements-advanced.txt
```

### Conda installation

```bash
conda env create -f environment.yml
conda activate ds-ml-dl-master-course

```

## Reproducibility rules

1. Start Jupyter from the extracted course root.
2. Keep raw bundled data unchanged.
3. Write generated files to `artifacts/`.
4. Restart the kernel and run all cells before considering a notebook complete.
5. Record package versions, seeds, split rules, and important configuration.
6. Never fit preprocessing on a complete dataset before validation.

## Troubleshooting

- **`ModuleNotFoundError`:** activate the intended environment and install the matching requirements file.
- **Dataset not found:** start Jupyter inside the extracted course folder; do not move individual notebooks alone.
- **Variables missing:** restart the kernel and use **Run All**.
- **Binary/package conflicts:** create a clean Python 3.11 environment instead of patching a heavily modified global environment.
- **PyTorch unavailable:** complete the core course through Phase 8 and install `requirements-advanced.txt` before Phase 9.
- **Plots do not appear:** verify the notebook is trusted and the Matplotlib cells complete without an earlier exception.

Run `python tools/verify_course.py` from the course root for structural validation.
