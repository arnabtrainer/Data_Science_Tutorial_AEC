````python
from sklearn.cluster import AgglomerativeClustering, DBSCAN
from sklearn.metrics import silhouette_score
from sklearn.preprocessing import StandardScaler

df=pd.read_csv(DATA_DIR/"customer_segments.csv")
X=StandardScaler().fit_transform(df.drop(columns="customer_id"))
agg=AgglomerativeClustering(n_clusters=4).fit_predict(X)
db=DBSCAN(eps=.65,min_samples=8).fit_predict(X)
print("Agglomerative silhouette:",silhouette_score(X,agg))
valid=db!=-1
print("DBSCAN clusters:",len(set(db))-(-1 in db),"noise:",int((~valid).sum()))
if len(set(db[valid]))>1:
    print("DBSCAN non-noise silhouette:",silhouette_score(X[valid],db[valid]))
````
