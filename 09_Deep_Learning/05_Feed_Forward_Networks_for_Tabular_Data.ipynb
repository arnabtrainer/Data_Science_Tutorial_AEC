## 7. Worked Python example

````python
import torch
from torch import nn
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler

df=pd.read_csv(DATA_DIR/"student_scores.csv")
X=df[["study_hours_per_week","attendance_pct","sleep_hours","prior_score"]].to_numpy(np.float32)
y=df["passed"].to_numpy(np.int64)
Xtr,Xte,ytr,yte=train_test_split(X,y,test_size=.25,stratify=y,random_state=42)
scaler=StandardScaler(); Xtr=scaler.fit_transform(Xtr); Xte=scaler.transform(Xte)
Xtr=torch.tensor(Xtr,dtype=torch.float32); Xte=torch.tensor(Xte,dtype=torch.float32)
ytr=torch.tensor(ytr); yte=torch.tensor(yte)
model=nn.Sequential(nn.Linear(4,16),nn.ReLU(),nn.Linear(16,8),nn.ReLU(),nn.Linear(8,2))
opt=torch.optim.Adam(model.parameters(),lr=.01,weight_decay=1e-4)
loss_fn=nn.CrossEntropyLoss()
for _ in range(150):
    opt.zero_grad(); loss=loss_fn(model(Xtr),ytr); loss.backward(); opt.step()
with torch.no_grad():
    accuracy=(model(Xte).argmax(1)==yte).float().mean().item()
print("Test accuracy:",accuracy)
````

## 10. Common mistakes and safeguards

        | Common mistake | Professional safeguard |
        |---|---|
        | Debugging architecture before data | Verify labels, batches, shapes, and a tiny overfit test first. |
