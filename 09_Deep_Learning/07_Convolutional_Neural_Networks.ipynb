## 7. Worked Python example

Read the code once without running it. Predict:

- the important intermediate objects and their shapes/types;

- one change that would cause a failure or misleading result.

Then run the cell and reconcile the actual output with your prediction.

````python
import torch
from torch import nn
from torch.utils.data import DataLoader, TensorDataset
torch.manual_seed(42)
X=np.load(DATA_DIR/"images/simple_shapes_X.npy")
y=np.load(DATA_DIR/"images/simple_shapes_y.npy")
order=np.random.default_rng(42).permutation(len(X))
split=int(.8*len(X)); train_idx,test_idx=order[:split],order[split:]
train_loader=DataLoader(TensorDataset(torch.tensor(X[train_idx]),torch.tensor(y[train_idx])),batch_size=48,shuffle=True)
Xtest=torch.tensor(X[test_idx]); ytest=torch.tensor(y[test_idx])
model=nn.Sequential(
    nn.Conv2d(1,8,3,padding=1),nn.ReLU(),nn.MaxPool2d(2),
    nn.Conv2d(8,16,3,padding=1),nn.ReLU(),nn.MaxPool2d(2),
    nn.Flatten(),nn.Linear(16*4*4,3)
)
opt=torch.optim.Adam(model.parameters(),lr=.005); loss_fn=nn.CrossEntropyLoss()
for epoch in range(8):
    for xb,yb in train_loader:
        opt.zero_grad(); loss=loss_fn(model(xb),yb); loss.backward(); opt.step()
with torch.no_grad():
    acc=(model(Xtest).argmax(1)==ytest).float().mean().item()
print("Shape-classification test accuracy:",acc)
````

## 10. Common mistakes and safeguards

        | Common mistake | Professional safeguard |
        |---|---|
        | Debugging architecture before data | Verify labels, batches, shapes, and a tiny overfit test first. |

| Forgetting train/eval modes | Switch modes for dropout and normalization. |
